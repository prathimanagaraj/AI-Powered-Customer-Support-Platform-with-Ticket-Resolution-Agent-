```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

# Keep only 200 rows
df = df.head(200)

# Step 1: Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Cabin'] = df['Cabin'].fillna('Unknown')
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("Rows and Columns:", df.shape)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
print("Data types checked successfully")
print(df.dtypes)
```

```python
df.describe()
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[(df['Fare'] < lower_limit) | (df['Fare'] > upper_limit)]

print("Number of Fare outliers:", len(outliers))
```

```python
df = df[(df['Fare'] >= lower_limit) & (df['Fare'] <= upper_limit)]

print("Rows after removing outliers:", len(df))
```

```python
print(df['Fare'].max())
```

```python
df['Name'] = df['Name'].str.strip()

print(df['Name'].head())
```

```python
df['Sex'] = df['Sex'].str.strip().str.lower()

print(df['Sex'].unique())
```

```python
df['Embarked'] = df['Embarked'].str.strip().str.upper()

print(df['Embarked'].unique())
```

```python
df_encoded = pd.get_dummies(df, columns=['Sex', 'Embarked'])

print(df_encoded.head())
```
